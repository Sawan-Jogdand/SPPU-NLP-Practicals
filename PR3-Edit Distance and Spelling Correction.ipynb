```python
from collections import Counter
```

```python
def levenshtain_distance(word1, word2):

    m = len(word1)
    n = len(word2)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i

    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):

            if word1[i - 1] == word2[j - 1]:
                cost = 0
            else:
                cost = 1

            dp[i][j] = min(
                dp[i - 1][j] + 1,       
                dp[i][j - 1] + 1,      
                dp[i - 1][j - 1] + cost 
            )

    return dp[m][n]
```

```python
word1 = "om"
word2 = "raj"

distance = levenshtain_distance(word1, word2)

print("Word 1:", word1)
print("Word 2:", word2)
print("Edit distance is:", distance)
```

```python
dictionary = [
    "receive",
    "recipe",
    "relieve",
    "receiver",
    "computer",
    "language"
]
```

```python
def find_closest_words(misspelled_word, dictionary):
    distances = []

    for word in dictionary:
        distance = levenshtain_distance(misspelled_word, word)
        distances.append((word, distance))

    distances.sort(key=lambda x: x[1])

    return distances
```

```python
misspelled_word = "recieve"

results = find_closest_words(misspelled_word, dictionary)

print("Misspelled word:", misspelled_word)
print("\nCandidate words:")

for word, distance in results:
    print(word, "->", distance)
```
