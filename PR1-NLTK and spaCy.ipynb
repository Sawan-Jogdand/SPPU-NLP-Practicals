```python
#Tokenization
import nltk

from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

text = "Natural Language Processing is an interesting field."

tokens = word_tokenize(text)

print("Original Text:")
print(text)

print("\nTokens:")
print(tokens)
```

```python
#Stop-word Removal
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = "Natural Language Processing is an interesting field."

tokens = word_tokenize(text.lower())

stop_words = set(stopwords.words("english"))

result = []

for word in tokens:
    if word not in stop_words:
        result.append(word)

print("Original Tokens:")
print(tokens)

print("\nAfter Stop-word Removal:")
print(result)
```

```python
#Lemmatization
import nltk

from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "studies", "cars", "running"]

print("Original Words:")
print(words)

print("\nLemmatized Words:")

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

```python
#Stemming
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "studies", "connected", "running"]

print("Original Words:")
print(words)

print("\nStemmed Words:")

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
#POS Tagging using NLTK
import nltk

from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

text = "The boy is playing football."

tokens = word_tokenize(text)

pos_tags = nltk.pos_tag(tokens)

print("POS Tags:")

for word, tag in pos_tags:
    print(word, "->", tag)
```

```python
#Morphological Analysis
words = ["playing", "played", "player", "happiness", "quickly"]

suffixes = ["ing", "ed", "er", "ness", "ly", "s"]

for word in words:

    root = word
    suffix = ""

    for s in suffixes:
        if word.endswith(s):
            root = word[:-len(s)]
            suffix = s
            break

    print("Word:", word)
    print("Root:", root)
    print("Suffix:", suffix)
    print()
```

```python
#Ambiguity – Lexical
sentence1 = "I went to the bank to deposit money."
sentence2 = "I sat near the bank of the river."

print("Sentence 1:")
print(sentence1)

print("\nAmbiguous Word: bank")
print("Meaning: Financial institution")

print("\nSentence 2:")
print(sentence2)

print("\nAmbiguous Word: bank")
print("Meaning: Side of a river")
```

```python
#Ambiguity – Syntactic

sentence = "I saw the man with a telescope."

print("Sentence:")
print(sentence)

print("\nSyntactic Ambiguity:")

print("Meaning 1: I used a telescope to see the man.")

print("Meaning 2: The man had a telescope.")
```

```python
#POS Tagging using spaCy
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The boy is playing football."

doc = nlp(text)

print("POS Tags using spaCy:")

for token in doc:
    print(token.text, "->", token.pos_)
```
