```python
import re

text = "Today we learned about #Python, #NLP, #MachineLearning, and #AI.Follow #OpenAI for more updates!"

pattern = r'#\w+'

hash = re.findall(pattern , text)

print("extracted Hashtags ")

for tag in hash:
    print(tag)
```

```python
def dfa(binary):
    state = "q0"

    for bit in binary:

        if state == "q0":
            if bit == "0":
                state = "q1"
            else:
                state = "q0"

        elif state == "q1":
            if bit == "0":
                state = "q1"
            else:
                state = "q2"

        elif state == "q2":
            if bit == "0":
                state = "q1"
            else:
                state = "q0"

    if state == "q2":
        print("Accepted")
    else:
        print("Rejected")


binary = input("Enter binary string: ")
dfa(binary)
```

```python
# NFA 

def nfa(binary):

    # Step 1: Start in the initial state q0
    current_states = {"q0"}

    # Step 2: Read each bit from the input string
    for bit in binary:

        # Store all possible next states
        next_states = set()

        # Process every current state
        for state in current_states:

            # Transition rules for q0
            if state == "q0":

                if bit == "0":
                    # NFA has two choices
                    next_states.add("q0")
                    next_states.add("q1")

                elif bit == "1":
                    next_states.add("q0")

            # Transition rules for q1
            elif state == "q1":

                if bit == "1":
                    next_states.add("q2")

            # q2 is the accepting state
            elif state == "q2":
                pass

        # Move to the newly found states
        current_states = next_states

    # Check if the accepting state is reached
    if "q2" in current_states:
        print("Accepted")
    else:
        print("Rejected")


binary = input("Enter a binary string: ")
nfa(binary)
```
